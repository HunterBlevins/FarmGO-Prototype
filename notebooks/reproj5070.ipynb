Reprojects a folder of rasters to 5070 for later zonal stats calculations

```python
import arcpy
import os
```

```python
rasterfolder = r""

output_root = os.path.join(rasterfolder, "XYZ_5070") #Replace with proper folder name

# EPSG:5070
target_crs = arcpy.SpatialReference(5070)

# Create output root
os.makedirs(output_root, exist_ok=True)

# ------------------------------------------------------------
# WALK THROUGH ALL SUBFOLDERS
# ------------------------------------------------------------

for root, dirs, files in os.walk(rasterfolder):

    # Don't process the output folder itself
    if os.path.abspath(root).startswith(os.path.abspath(output_root)):
        continue

    for file in files:

        # Only process raster files
        if not file.lower().endswith((
            ".tif",
            ".tiff",
            ".img",
            ".crf"
        )):
            continue

        input_raster = os.path.join(root, file)

        # Determine path relative to original Prism folder
        relative_path = os.path.relpath(
            input_raster,
            rasterfolder
        )

        # Output location preserving folder structure
        output_raster = os.path.join(
            output_root,
            relative_path
        )

        # Create the required output subfolder
        output_folder = os.path.dirname(output_raster)
        os.makedirs(output_folder, exist_ok=True)

        print(f"Projecting:")
        print(f"  Input:  {input_raster}")
        print(f"  Output: {output_raster}")

        # ----------------------------------------------------
        # PROJECT RASTER
        # ----------------------------------------------------

        arcpy.management.ProjectRaster(
            in_raster=input_raster,
            out_raster=output_raster,
            out_coor_system=target_crs,
            resampling_type="BILINEAR"
        )

print("Finished projecting all rasters to EPSG:5070.")
```
