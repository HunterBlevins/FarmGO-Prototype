Downloads PRISM data into a folder within an ArcGIS project.
Documentation - https://prism.oregonstate.edu/documents/PRISM_downloads_web_service.pdf

```python
import arcpy
import os
import requests
import zipfile
import io
```

```python
# -------------------------
# ACTIVE ARCGIS PROJECT
# -------------------------
project = arcpy.mp.ArcGISProject("CURRENT")
project_folder = os.path.dirname(project.filePath)

# Main downloaded data folder
downloaded_data_folder = os.path.join(project_folder, "Downloaded Data")
os.makedirs(downloaded_data_folder, exist_ok=True)

# PRISM folder
prism_folder = os.path.join(downloaded_data_folder, "Prism")
os.makedirs(prism_folder, exist_ok=True)

# -------------------------
# SETTINGS
# -------------------------
region = "us"

variables = [
    "ppt",      # precipitation
    "tmin",     # minimum temperature
    "tmax",     # maximum temperature
    "tmean",    # mean temperature
    "tdmean"    # mean dewpoint
]

resolution = "800m"
time = "202608"       # YYYYMM (monthly), YYYYMMDD (daily), YYYY (annual)

# -------------------------
# DOWNLOAD
# -------------------------
for variable in variables:

    # Create folder for this variable
    variable_folder = os.path.join(prism_folder, variable)
    os.makedirs(variable_folder, exist_ok=True)

    # Build URL
    url = f"https://services.nacse.org/prism/data/get/{region}/{resolution}/{variable}/{time}"

    print()
    print("=" * 60)
    print(f"Downloading {variable}")
    print(url)

    # Download
    r = requests.get(url)
    r.raise_for_status()

    # Extract into variable folder
    z = zipfile.ZipFile(io.BytesIO(r.content))
    z.extractall(variable_folder)

    print(f"Done: {variable}")
    print(f"Saved to: {variable_folder}")

print()
print("All PRISM downloads complete.")
```
